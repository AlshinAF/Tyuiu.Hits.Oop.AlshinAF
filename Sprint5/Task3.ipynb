```
using System;
using System.Collections.Generic;
using System.Linq;

// [Все интерфейсы, классы и делегаты остаются без изменений - вставляем их здесь]
// Интерфейсы
public interface IIdentifiable
{
    string GetId();
    string GetEntityType();
    DateTime GetCreatedDate();
    bool ValidateId();
}

public interface INotificationService
{
    void SendNotification(string message, Person recipient);
    void SendBulkNotification(string message, List<Person> recipients);
}

public interface IAnalyticsService
{
    void TrackPersonCreation(Person person);
    void TrackInteraction(string interactionType, Person person1, Person person2);
    void GenerateReport();
}

// Реализация сервисов
public class EmailNotificationService : INotificationService
{
    public void SendNotification(string message, Person recipient)
    {
        Console.WriteLine($"📧 Email отправлен на {recipient.Email}: {message}");
    }

    public void SendBulkNotification(string message, List<Person> recipients)
    {
        Console.WriteLine($"📧 Массовая рассылка на {recipients.Count} человек: {message}");
        foreach (var recipient in recipients.Take(3))
        {
            Console.WriteLine($"   - {recipient.Name} ({recipient.Email})");
        }
        if (recipients.Count > 3)
            Console.WriteLine($"   ... и еще {recipients.Count - 3} человек");
    }
}

public class AnalyticsService : IAnalyticsService
{
    private List<string> _events = new List<string>();

    public void TrackPersonCreation(Person person)
    {
        string eventInfo = $"[{DateTime.Now:HH:mm:ss}] Создан {person.GetType().Name}: {person.Name}";
        _events.Add(eventInfo);
        Console.WriteLine($"📊 Аналитика: {eventInfo}");
    }

    public void TrackInteraction(string interactionType, Person person1, Person person2)
    {
        string eventInfo = $"[{DateTime.Now:HH:mm:ss}] Взаимодействие '{interactionType}' между {person1.Name} и {person2.Name}";
        _events.Add(eventInfo);
        Console.WriteLine($"📊 Аналитика: {eventInfo}");
    }

    public void GenerateReport()
    {
        Console.WriteLine("\n=== ОТЧЕТ АНАЛИТИКИ ===");
        Console.WriteLine($"Всего событий: {_events.Count}");
        foreach (var eventItem in _events.TakeLast(10))
        {
            Console.WriteLine($"  • {eventItem}");
        }
    }
}

// Контейнер зависимостей
public class DependencyContainer
{
    private static Dictionary<Type, object> _services = new Dictionary<Type, object>();

    public static void Register<T>(T service)
    {
        _services[typeof(T)] = service;
    }

    public static T Resolve<T>()
    {
        if (_services.ContainsKey(typeof(T)))
        {
            return (T)_services[typeof(T)];
        }
        throw new InvalidOperationException($"Сервис {typeof(T).Name} не зарегистрирован");
    }
}

// Делегаты для системы событий
public delegate void LanguageEventHandler(object sender, LanguageEventArgs e);
public delegate void IncomeChangedEventHandler(object sender, IncomeChangedEventArgs e);
public delegate void PersonActionEventHandler(object sender, PersonActionEventArgs e);

// Аргументы событий
public class LanguageEventArgs : EventArgs
{
    public string Language { get; set; }
    public string ActionType { get; set; }
    
    public LanguageEventArgs(string language, string actionType)
    {
        Language = language;
        ActionType = actionType;
    }
}

public class IncomeChangedEventArgs : EventArgs
{
    public decimal OldIncome { get; set; }
    public decimal NewIncome { get; set; }
    public string Reason { get; set; }
    
    public IncomeChangedEventArgs(decimal oldIncome, decimal newIncome, string reason)
    {
        OldIncome = oldIncome;
        NewIncome = newIncome;
        Reason = reason;
    }
}

public class PersonActionEventArgs : EventArgs
{
    public string ActionType { get; set; }
    public string Description { get; set; }
    public DateTime Timestamp { get; set; }
    
    public PersonActionEventArgs(string actionType, string description)
    {
        ActionType = actionType;
        Description = description;
        Timestamp = DateTime.Now;
    }
}

// Базовый класс Person с событиями
public class Person : IIdentifiable
{
    // Существующие свойства
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }
    public string Email { get; set; }
    public string PhoneNumber { get; set; }
    public bool IsMarried { get; set; }
    public string Nationality { get; set; }
    public DateTime RegistrationDate { get; set; }
    public string Address { get; set; }
    public string EmergencyContact { get; set; }
    public List<string> Languages { get; set; }
    public decimal MonthlyIncome { get; set; }

    // Новые атрибуты с коллекциями
    public List<string> Skills { get; set; }
    public Dictionary<string, string> ContactPreferences { get; set; }
    public HashSet<string> CompletedTrainings { get; set; }
    public Queue<string> MessageQueue { get; set; }

    // События
    public event LanguageEventHandler LanguageAdded;
    public event LanguageEventHandler LanguageRemoved;
    public event IncomeChangedEventHandler IncomeChanged;
    public event PersonActionEventHandler PersonActionPerformed;

    // Явная реализация интерфейса IIdentifiable
    string IIdentifiable.GetId()
    {
        return $"PERSON_{Name.Replace(" ", "_").ToUpper()}_{RegistrationDate:yyyyMMdd}";
    }

    string IIdentifiable.GetEntityType() => "Person";
    DateTime IIdentifiable.GetCreatedDate() => RegistrationDate;

    bool IIdentifiable.ValidateId()
    {
        var identifiable = (IIdentifiable)this;
        return !string.IsNullOrEmpty(identifiable.GetId()) && 
               identifiable.GetId().StartsWith("PERSON_");
    }

    public Person() 
    {
        RegistrationDate = DateTime.Now;
        Languages = new List<string>();
        Skills = new List<string>();
        ContactPreferences = new Dictionary<string, string>();
        CompletedTrainings = new HashSet<string>();
        MessageQueue = new Queue<string>();
        InitializeDefaultContactPreferences();
    }

    public Person(string name, int age, string gender, string email, string phoneNumber, bool isMarried)
        : this()
    {
        Name = name;
        Age = age;
        Gender = gender;
        Email = email;
        PhoneNumber = phoneNumber;
        IsMarried = isMarried;
    }

    private void InitializeDefaultContactPreferences()
    {
        ContactPreferences["Email"] = "Enabled";
        ContactPreferences["SMS"] = "Enabled";
        ContactPreferences["Push"] = "Disabled";
        ContactPreferences["Phone"] = "EmergencyOnly";
    }

    protected virtual void OnLanguageAdded(string language)
    {
        LanguageAdded?.Invoke(this, new LanguageEventArgs(language, "Added"));
    }

    protected virtual void OnLanguageRemoved(string language)
    {
        LanguageRemoved?.Invoke(this, new LanguageEventArgs(language, "Removed"));
    }

    protected virtual void OnIncomeChanged(decimal oldIncome, decimal newIncome, string reason)
    {
        IncomeChanged?.Invoke(this, new IncomeChangedEventArgs(oldIncome, newIncome, reason));
    }

    protected virtual void OnPersonActionPerformed(string actionType, string description)
    {
        PersonActionPerformed?.Invoke(this, new PersonActionEventArgs(actionType, description));
    }

    // Виртуальные методы для переопределения
    public virtual void AddLanguage(string language)
    {
        if (!Languages.Contains(language))
        {
            Languages.Add(language);
            OnLanguageAdded(language);
            Console.WriteLine($"{Name} выучил(а) язык: {language}");
        }
    }

    public virtual void RemoveLanguage(string language)
    {
        if (Languages.Remove(language))
        {
            OnLanguageRemoved(language);
            Console.WriteLine($"{Name} забыл(а) язык: {language}");
        }
    }

    public virtual void RequestTimeOff(int days, string reason)
    {
        Console.WriteLine($"{Name} запрашивает отпуск на {days} дней. Причина: {reason}");
        
        try
        {
            var notificationService = DependencyContainer.Resolve<INotificationService>();
            notificationService.SendNotification($"Запрос отпуска от {Name} на {days} дней: {reason}", this);
        }
        catch (InvalidOperationException)
        {
            Console.WriteLine("Сервис уведомлений недоступен");
        }
    }

    public virtual void CalculateAnnualIncome()
    {
        decimal annualIncome = MonthlyIncome * 12;
        Console.WriteLine($"{Name}: Годовой доход составляет {annualIncome:C}");
    }

    // Новые методы с обработкой коллекций
    public virtual void AddSkill(string skill)
    {
        if (!Skills.Contains(skill))
        {
            Skills.Add(skill);
            Skills.Sort();
            OnPersonActionPerformed("SkillAdded", $"Добавлен навык: {skill}");
            Console.WriteLine($"{Name} приобрел(а) навык: {skill}");
        }
    }

    public virtual void RemoveSkill(string skill)
    {
        if (Skills.Remove(skill))
        {
            OnPersonActionPerformed("SkillRemoved", $"Удален навык: {skill}");
            Console.WriteLine($"{Name} утратил(а) навык: {skill}");
        }
    }

    public void UpdateContactPreference(string channel, string preference)
    {
        if (ContactPreferences.ContainsKey(channel))
        {
            ContactPreferences[channel] = preference;
            OnPersonActionPerformed("ContactPreferenceChanged", 
                $"Изменены предпочтения для {channel}: {preference}");
        }
    }

    public void CompleteTraining(string trainingName)
    {
        if (CompletedTrainings.Add(trainingName))
        {
            OnPersonActionPerformed("TrainingCompleted", $"Завершено обучение: {trainingName}");
            Console.WriteLine($"{Name} завершил(а) обучение: {trainingName}");
        }
    }

    public void SendMessage(string message)
    {
        MessageQueue.Enqueue(message);
        OnPersonActionPerformed("MessageSent", $"Отправлено сообщение: {message}");
        
        if (MessageQueue.Count > 10)
        {
            MessageQueue.Dequeue(); // Удаляем старое сообщение
        }
    }

    public string GetNextMessage()
    {
        return MessageQueue.Count > 0 ? MessageQueue.Dequeue() : "Нет сообщений";
    }

    public virtual void UpdateIncome(decimal newIncome, string reason)
    {
        if (newIncome != MonthlyIncome)
        {
            decimal oldIncome = MonthlyIncome;
            MonthlyIncome = newIncome;
            OnIncomeChanged(oldIncome, newIncome, reason);
            Console.WriteLine($"{Name}: Доход изменен с {oldIncome:C} на {newIncome:C}. Причина: {reason}");
        }
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {Email}, Телефон: {PhoneNumber}, " +
               $"В браке: {(IsMarried ? "Да" : "Нет")}, Национальность: {Nationality}, " +
               $"Адрес: {Address}, Экстренный контакт: {EmergencyContact}, Месячный доход: {MonthlyIncome:C}";
    }

    public string GetLanguages()
    {
        return Languages.Count > 0 ? string.Join(", ", Languages) : "Языки не указаны";
    }

    public virtual void DisplaySkills()
    {
        Console.WriteLine($"{Name} имеет {Skills.Count} навыков:");
        foreach (var skill in Skills)
        {
            Console.WriteLine($"  - {skill}");
        }
    }

    public virtual void DisplayContactPreferences()
    {
        Console.WriteLine($"Предпочтения связи для {Name}:");
        foreach (var pref in ContactPreferences)
        {
            Console.WriteLine($"  {pref.Key}: {pref.Value}");
        }
    }
}

public class Student : Person, IIdentifiable
{
    // Существующие свойства
    public string University { get; set; }
    public int Course { get; set; }
    public List<string> Courses { get; set; }
    public string StudentId { get; set; }
    public double AverageGrade { get; set; }
    public string Faculty { get; set; }
    public bool HasScholarship { get; set; }
    public string Major { get; set; }
    public int CreditsCompleted { get; set; }
    public DateTime ExpectedGraduation { get; set; }
    public bool IsInternational { get; set; }
    public decimal ScholarshipAmount { get; set; }

    // Новые атрибуты с коллекциями
    public Dictionary<string, double> CourseGrades { get; set; }
    public List<string> ExtracurricularActivities { get; set; }
    public Stack<string> AssignmentStack { get; set; }
    public LinkedList<string> AcademicGoals { get; set; }

    // События специфичные для Student
    public event EventHandler CourseEnrolled;
    public event EventHandler GradeAdded;

    // Явная реализация интерфейса IIdentifiable
    string IIdentifiable.GetId() => StudentId ?? $"STUDENT_{Name.Replace(" ", "_").ToUpper()}_{University?.Replace(" ", "_").ToUpper()}";
    string IIdentifiable.GetEntityType() => "Student";
    bool IIdentifiable.ValidateId()
    {
        var identifiable = (IIdentifiable)this;
        return !string.IsNullOrEmpty(identifiable.GetId()) && 
               (identifiable.GetId().StartsWith("STUDENT_") || identifiable.GetId().StartsWith("S"));
    }

    public Student(string name, int age, string gender, string email, string phoneNumber, 
                   bool isMarried, string university, int course, string studentId, 
                   double averageGrade, string faculty = "Не указан", bool hasScholarship = false,
                   string major = "", int creditsCompleted = 0, bool isInternational = false,
                   decimal scholarshipAmount = 0) 
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        University = university;
        Course = course;
        StudentId = studentId;
        AverageGrade = averageGrade;
        Faculty = faculty;
        HasScholarship = hasScholarship;
        Major = major;
        CreditsCompleted = creditsCompleted;
        IsInternational = isInternational;
        ScholarshipAmount = scholarshipAmount;
        MonthlyIncome = scholarshipAmount;
        ExpectedGraduation = DateTime.Now.AddYears(4 - course);
        
        Courses = new List<string>();
        CourseGrades = new Dictionary<string, double>();
        ExtracurricularActivities = new List<string>();
        AssignmentStack = new Stack<string>();
        AcademicGoals = new LinkedList<string>();
    }

    protected virtual void OnCourseEnrolled()
    {
        CourseEnrolled?.Invoke(this, EventArgs.Empty);
    }

    protected virtual void OnGradeAdded()
    {
        GradeAdded?.Invoke(this, EventArgs.Empty);
    }

    // Новые методы
    public void AddCourseGrade(string course, double grade)
    {
        CourseGrades[course] = grade;
        UpdateAverageGrade();
        OnGradeAdded();
        OnPersonActionPerformed("GradeAdded", $"Оценка {grade} за курс {course}");
    }

    private void UpdateAverageGrade()
    {
        if (CourseGrades.Count > 0)
        {
            AverageGrade = CourseGrades.Values.Average();
        }
    }

    public void AddExtracurricularActivity(string activity)
    {
        if (!ExtracurricularActivities.Contains(activity))
        {
            ExtracurricularActivities.Add(activity);
            OnPersonActionPerformed("ActivityAdded", $"Добавлена внеучебная деятельность: {activity}");
        }
    }

    public void PushAssignment(string assignment)
    {
        AssignmentStack.Push(assignment);
        OnPersonActionPerformed("AssignmentPushed", $"Добавлено задание: {assignment}");
    }

    public string PopAssignment()
    {
        return AssignmentStack.Count > 0 ? AssignmentStack.Pop() : "Нет заданий";
    }

    public void AddAcademicGoal(string goal)
    {
        AcademicGoals.AddLast(goal);
        OnPersonActionPerformed("GoalAdded", $"Добавлена академическая цель: {goal}");
    }

    public void DisplayAcademicProgress()
    {
        Console.WriteLine($"\nАкадемический прогресс {Name}:");
        Console.WriteLine($"Завершено кредитов: {CreditsCompleted}/120");
        Console.WriteLine($"Средний балл: {AverageGrade:F2}");
        Console.WriteLine($"Курсов завершено: {CourseGrades.Count}");
        Console.WriteLine("Оценки по курсам:");
        foreach (var grade in CourseGrades)
        {
            Console.WriteLine($"  {grade.Key}: {grade.Value:F1}");
        }
    }

    // Специфичные методы Student
    public void AddCourse(string course)
    {
        if (!Courses.Contains(course))
        {
            Courses.Add(course);
            CreditsCompleted += 3;
            OnCourseEnrolled();
            OnPersonActionPerformed("CourseEnrolled", $"Зачислен на курс: {course}");
            Console.WriteLine($"{Name} добавлен в курс: {course}. Всего кредитов: {CreditsCompleted}");
        }
    }

    public void RemoveCourse(string course)
    {
        if (Courses.Remove(course))
        {
            CreditsCompleted = Math.Max(0, CreditsCompleted - 3);
            Console.WriteLine($"{Name} удален из курса: {course}. Всего кредитов: {CreditsCompleted}");
        }
    }

    public void CalculateGPA()
    {
        double gpa = AverageGrade / 5.0 * 4.0;
        Console.WriteLine($"{Name}: GPA составляет {gpa:F2} из 4.0");
    }

    public void ApplyForGraduation()
    {
        if (CreditsCompleted >= 120)
        {
            Console.WriteLine($"{Name} подает заявление на выпуск. Ожидаемая дата выпуска: {ExpectedGraduation:yyyy-MM-dd}");
        }
        else
        {
            Console.WriteLine($"{Name} не может выпуститься. Необходимо {120 - CreditsCompleted} дополнительных кредитов.");
        }
    }

    // Переопределенные методы
    public override void CalculateAnnualIncome()
    {
        decimal annualIncome = MonthlyIncome * 9; // Стипендия только 9 месяцев в году
        Console.WriteLine($"{Name}: Годовой доход (стипендия) составляет {annualIncome:C}");
    }

    public override void UpdateIncome(decimal newIncome, string reason)
    {
        base.UpdateIncome(newIncome, reason);
        ScholarshipAmount = newIncome; // Для студента доход - это стипендия
    }

    public override void RequestTimeOff(int days, string reason)
    {
        Console.WriteLine($"{Name} (студент) запрашивает академический отпуск на {days} дней. Причина: {reason}");
        
        try
        {
            var notificationService = DependencyContainer.Resolve<INotificationService>();
            notificationService.SendNotification($"Академический отпуск от {Name} на {days} дней: {reason}", this);
        }
        catch (InvalidOperationException)
        {
            Console.WriteLine("Сервис уведомлений недоступен");
        }
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Университет: {University}, Факультет: {Faculty}, " +
               $"Курс: {Course}, ID: {StudentId}, Средний балл: {AverageGrade:F1}, " +
               $"Стипендия: {(HasScholarship ? "Да" : "Нет")}, Специальность: {Major}, " +
               $"Завершено кредитов: {CreditsCompleted}, Международный: {(IsInternational ? "Да" : "Нет")}";
    }
}

public class Employee : Person, IIdentifiable
{
    // Существующие свойства
    public string Company { get; set; }
    public string Position { get; set; }
    public string EmployeeId { get; set; }
    public int VacationDays { get; set; }
    public string Department { get; set; }
    public int YearsInCompany { get; set; }
    public string Manager { get; set; }
    public string WorkSchedule { get; set; }
    public bool IsRemote { get; set; }
    public int ProjectsCompleted { get; set; }

    // Новые атрибуты с коллекциями
    public Dictionary<string, decimal> ProjectBudgets { get; set; }
    public List<DateTime> VacationRequests { get; set; }
    public Queue<string> TaskQueue { get; set; }
    public SortedDictionary<DateTime, string> Meetings { get; set; }

    // События
    public event EventHandler ProjectCompleted;
    public event EventHandler PromotionRequested;

    // Явная реализация интерфейса IIdentifiable
    string IIdentifiable.GetId() => EmployeeId ?? $"EMPLOYEE_{Name.Replace(" ", "_").ToUpper()}_{Company?.Replace(" ", "_").ToUpper()}";
    string IIdentifiable.GetEntityType() => "Employee";
    bool IIdentifiable.ValidateId()
    {
        var identifiable = (IIdentifiable)this;
        return !string.IsNullOrEmpty(identifiable.GetId()) && 
               (identifiable.GetId().StartsWith("EMPLOYEE_") || identifiable.GetId().StartsWith("E"));
    }

    public Employee(string name, int age, string gender, string email, string phoneNumber, 
                    bool isMarried, string company, decimal salary, string employeeId, 
                    int vacationDays, string department = "Не указан", int yearsInCompany = 0, 
                    string position = "Сотрудник", string manager = "", string workSchedule = "9-18",
                    bool isRemote = false, int projectsCompleted = 0) 
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        Company = company;
        Position = position;
        EmployeeId = employeeId;
        VacationDays = vacationDays;
        Department = department;
        YearsInCompany = yearsInCompany;
        Manager = manager;
        WorkSchedule = workSchedule;
        IsRemote = isRemote;
        ProjectsCompleted = projectsCompleted;
        MonthlyIncome = salary;
        
        ProjectBudgets = new Dictionary<string, decimal>();
        VacationRequests = new List<DateTime>();
        TaskQueue = new Queue<string>();
        Meetings = new SortedDictionary<DateTime, string>();
    }

    protected virtual void OnProjectCompleted()
    {
        ProjectCompleted?.Invoke(this, EventArgs.Empty);
    }

    protected virtual void OnPromotionRequested()
    {
        PromotionRequested?.Invoke(this, EventArgs.Empty);
    }

    // Новые методы
    public void AddProjectWithBudget(string projectName, decimal budget)
    {
        ProjectBudgets[projectName] = budget;
        OnPersonActionPerformed("ProjectAdded", $"Добавлен проект {projectName} с бюджетом {budget:C}");
    }

    public void RequestVacation(DateTime date)
    {
        if (VacationDays > 0)
        {
            VacationRequests.Add(date);
            VacationDays--;
            OnPersonActionPerformed("VacationRequested", $"Запрошен отпуск на {date:yyyy-MM-dd}");
        }
    }

    public void AddTask(string task)
    {
        TaskQueue.Enqueue(task);
        OnPersonActionPerformed("TaskAdded", $"Добавлена задача: {task}");
    }

    public string ProcessNextTask()
    {
        return TaskQueue.Count > 0 ? TaskQueue.Dequeue() : "Нет задач в очереди";
    }

    public void ScheduleMeeting(DateTime date, string topic)
    {
        Meetings[date] = topic;
        OnPersonActionPerformed("MeetingScheduled", $"Запланирована встреча на {date:yyyy-MM-dd}: {topic}");
    }

    public void DisplayWorkSummary()
    {
        Console.WriteLine($"\nРабочая сводка {Name}:");
        Console.WriteLine($"Завершено проектов: {ProjectsCompleted}");
        Console.WriteLine($"Осталось отпускных дней: {VacationDays}");
        Console.WriteLine($"Задач в очереди: {TaskQueue.Count}");
        Console.WriteLine($"Запланировано встреч: {Meetings.Count}");
        Console.WriteLine("Текущие проекты и бюджеты:");
        foreach (var project in ProjectBudgets)
        {
            Console.WriteLine($"  {project.Key}: {project.Value:C}");
        }
    }

    // Специфичные методы Employee
    public void CompleteProject(string projectName)
    {
        ProjectsCompleted++;
        if (ProjectBudgets.ContainsKey(projectName))
        {
            ProjectBudgets.Remove(projectName);
        }
        OnProjectCompleted();
        OnPersonActionPerformed("ProjectCompleted", $"Завершен проект: {projectName}");
        
        try
        {
            var analyticsService = DependencyContainer.Resolve<IAnalyticsService>();
            analyticsService.TrackInteraction("project_completion", this, this);
        }
        catch (InvalidOperationException) { }
    }

    public void RequestPromotion(string desiredPosition)
    {
        Console.WriteLine($"{Name} запрашивает повышение до должности: {desiredPosition}");
        OnPromotionRequested();
        OnPersonActionPerformed("PromotionRequested", $"Запрос повышения до {desiredPosition}");
        
        try
        {
            var notificationService = DependencyContainer.Resolve<INotificationService>();
            notificationService.SendNotification($"Запрос повышения от {Name} на должность {desiredPosition}", this);
        }
        catch (InvalidOperationException)
        {
            Console.WriteLine("Сервис уведомлений недоступен");
        }
    }

    public void ScheduleMeeting(string withPerson, DateTime date, string topic)
    {
        Console.WriteLine($"{Name} назначает встречу с {withPerson} на {date:yyyy-MM-dd HH:mm}. Тема: {topic}");
    }

    public void RequestSalaryIncrease(decimal amount)
    {
        Console.WriteLine($"{Name} запрашивает увеличение зарплаты на {amount:C}");
        UpdateIncome(MonthlyIncome + amount, "Запрос на повышение зарплаты");
    }

    // Переопределенные методы
    public override void CalculateAnnualIncome()
    {
        decimal annualSalary = MonthlyIncome * 12;
        decimal bonus = annualSalary * 0.1m * YearsInCompany;
        decimal total = annualSalary + bonus;
        
        Console.WriteLine($"{Name}: Годовая зарплата {annualSalary:C} + бонус {bonus:C} = {total:C}");
    }

    public override void RequestTimeOff(int days, string reason)
    {
        Console.WriteLine($"{Name} (сотрудник) запрашивает отпуск на {days} дней. Причина: {reason}");
        
        if (days > VacationDays)
        {
            Console.WriteLine($"⚠️  Недостаточно отпускных дней. Доступно: {VacationDays}, Запрошено: {days}");
        }
        
        try
        {
            var notificationService = DependencyContainer.Resolve<INotificationService>();
            notificationService.SendNotification($"Запрос отпуска сотрудника {Name} на {days} дней: {reason}", this);
        }
        catch (InvalidOperationException)
        {
            Console.WriteLine("Сервис уведомлений недоступен");
        }
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Департамент: {Department}, " +
               $"Должность: {Position}, ID: {EmployeeId}, " +
               $"Лет в компании: {YearsInCompany}, Отпускных дней: {VacationDays}, " +
               $"Менеджер: {Manager}, График: {WorkSchedule}, Удаленная работа: {(IsRemote ? "Да" : "Нет")}, " +
               $"Завершено проектов: {ProjectsCompleted}";
    }
}

public class Teacher : Person, IIdentifiable
{
    // Существующие свойства
    public string Subject { get; set; }
    public int Experience { get; set; }
    public List<Student> Students { get; set; }
    public string TeacherId { get; set; }
    public string OfficeNumber { get; set; }
    public string AcademicDegree { get; set; }
    public int PublicationsCount { get; set; }
    public string ResearchArea { get; set; }
    public int OfficeHours { get; set; }
    public bool IsTenured { get; set; }
    public List<string> TeachingCertifications { get; set; }

    // Новые атрибуты с коллекциями
    public Dictionary<string, List<string>> CourseMaterials { get; set; }
    public List<DateTime> ResearchDeadlines { get; set; }
    public Stack<string> PublicationStack { get; set; }
    public Dictionary<Student, string> StudentEvaluations { get; set; }

    // События
    public event EventHandler ResearchPublished;
    public event EventHandler StudentMentored;

    // Явная реализация интерфейса IIdentifiable
    string IIdentifiable.GetId() => TeacherId ?? $"TEACHER_{Name.Replace(" ", "_").ToUpper()}_{Subject?.Replace(" ", "_").ToUpper()}";
    string IIdentifiable.GetEntityType() => "Teacher";
    bool IIdentifiable.ValidateId()
    {
        var identifiable = (IIdentifiable)this;
        return !string.IsNullOrEmpty(identifiable.GetId()) && 
               (identifiable.GetId().StartsWith("TEACHER_") || identifiable.GetId().StartsWith("T"));
    }

    public Teacher(string name, int age, string gender, string email, string phoneNumber, 
                   bool isMarried, string subject, int experience, string teacherId, 
                   string officeNumber, string academicDegree = "Кандидат наук", 
                   int publicationsCount = 0, string researchArea = "", int officeHours = 10,
                   bool isTenured = false, decimal salary = 0)
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        Subject = subject;
        Experience = experience;
        TeacherId = teacherId;
        OfficeNumber = officeNumber;
        AcademicDegree = academicDegree;
        PublicationsCount = publicationsCount;
        ResearchArea = researchArea;
        OfficeHours = officeHours;
        IsTenured = isTenured;
        MonthlyIncome = salary;
        
        TeachingCertifications = new List<string>();
        Students = new List<Student>();
        CourseMaterials = new Dictionary<string, List<string>>();
        ResearchDeadlines = new List<DateTime>();
        PublicationStack = new Stack<string>();
        StudentEvaluations = new Dictionary<Student, string>();
    }

    protected virtual void OnResearchPublished()
    {
        ResearchPublished?.Invoke(this, EventArgs.Empty);
    }

    protected virtual void OnStudentMentored()
    {
        StudentMentored?.Invoke(this, EventArgs.Empty);
    }

    // Новые методы
    public void AddCourseMaterial(string course, string material)
    {
        if (!CourseMaterials.ContainsKey(course))
        {
            CourseMaterials[course] = new List<string>();
        }
        CourseMaterials[course].Add(material);
        OnPersonActionPerformed("MaterialAdded", $"Добавлен материал для курса {course}: {material}");
    }

    public void AddResearchDeadline(DateTime deadline)
    {
        ResearchDeadlines.Add(deadline);
        ResearchDeadlines.Sort();
        OnPersonActionPerformed("DeadlineAdded", $"Добавлен дедлайн исследования: {deadline:yyyy-MM-dd}");
    }

    public void PushPublication(string publication)
    {
        PublicationStack.Push(publication);
        PublicationsCount++;
        OnResearchPublished();
        OnPersonActionPerformed("PublicationPushed", $"Добавлена публикация: {publication}");
    }

    public string PopPublication()
    {
        return PublicationStack.Count > 0 ? PublicationStack.Pop() : "Нет публикаций";
    }

    public void AddStudentEvaluation(Student student, string evaluation)
    {
        StudentEvaluations[student] = evaluation;
        OnPersonActionPerformed("EvaluationAdded", $"Добавлена оценка для студента {student.Name}");
    }

    public void DisplayTeachingPortfolio()
    {
        Console.WriteLine($"\nПреподавательское портфолио {Name}:");
        Console.WriteLine($"Количество студентов: {Students.Count}");
        Console.WriteLine($"Публикации: {PublicationsCount}");
        Console.WriteLine($"Курсы с материалами: {CourseMaterials.Count}");
        Console.WriteLine($"Предстоящие дедлайны: {ResearchDeadlines.Count}");
    }

    // Специфичные методы Teacher
    public void AddCertification(string certification)
    {
        if (!TeachingCertifications.Contains(certification))
        {
            TeachingCertifications.Add(certification);
            Console.WriteLine($"{Name} получил(а) сертификат: {certification}");
        }
    }

    public void ScheduleOfficeHours(string schedule)
    {
        OfficeHours = 10;
        Console.WriteLine($"{Name} установил(а) часы приема: {schedule}");
    }

    public void MentorStudent(Student student)
    {
        if (!Students.Contains(student))
        {
            Students.Add(student);
            OnStudentMentored();
            OnPersonActionPerformed("StudentMentored", $"Стал ментором для {student.Name}");
            Console.WriteLine($"{Name} стал(а) ментором для студента {student.Name}");
            
            try
            {
                var analyticsService = DependencyContainer.Resolve<IAnalyticsService>();
                analyticsService.TrackInteraction("mentoring", this, student);
            }
            catch (InvalidOperationException) { }
        }
    }

    public void SubmitResearchProposal(string title, string fundingAgency)
    {
        Console.WriteLine($"{Name} подает исследовательское предложение: '{title}' в {fundingAgency}");
        PushPublication($"Проект: {title} - {fundingAgency}");
        AddResearchDeadline(DateTime.Now.AddMonths(3));
    }

    // Переопределенные методы
    public override void CalculateAnnualIncome()
    {
        decimal baseAnnualIncome = MonthlyIncome * 12;
        decimal researchBonus = PublicationsCount * 5000; // Бонус за публикации
        decimal total = baseAnnualIncome + researchBonus;
        
        Console.WriteLine($"{Name}: Годовая зарплата {baseAnnualIncome:C} + бонус за публикации {researchBonus:C} = {total:C}");
    }

    public override void RequestTimeOff(int days, string reason)
    {
        Console.WriteLine($"{Name} (преподаватель) запрашивает академический отпуск на {days} дней. Причина: {reason}");
        
        if (days > 30)
        {
            Console.WriteLine("⚠️  Длительный отпуск требует одобрения декана");
        }
        
        try
        {
            var notificationService = DependencyContainer.Resolve<INotificationService>();
            notificationService.SendNotification($"Академический отпуск преподавателя {Name} на {days} дней: {reason}", this);
        }
        catch (InvalidOperationException)
        {
            Console.WriteLine("Сервис уведомлений недоступен");
        }
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Опыт: {Experience} лет, " +
               $"ID: {TeacherId}, Кабинет: {OfficeNumber}, Ученая степень: {AcademicDegree}, " +
               $"Публикации: {PublicationsCount}, Область исследований: {ResearchArea}, " +
               $"Часы приема: {OfficeHours}, Тенура: {(IsTenured ? "Да" : "Нет")}, " +
               $"Сертификаты: {TeachingCertifications.Count}";
    }
}

// Обновленный реестр с поддержкой событий
public class PersonRegistry<T> where T : Person
{
    private List<T> _people = new List<T>();
    private INotificationService _notificationService;
    private IAnalyticsService _analyticsService;

    // Событие реестра
    public event EventHandler<PersonEventArgs> PersonAdded;
    public event EventHandler<PersonEventArgs> PersonActionLogged;

    public class PersonEventArgs : EventArgs
    {
        public T Person { get; set; }
        public string Action { get; set; }
        
        public PersonEventArgs(T person, string action)
        {
            Person = person;
            Action = action;
        }
    }

    public PersonRegistry(INotificationService notificationService, IAnalyticsService analyticsService)
    {
        _notificationService = notificationService;
        _analyticsService = analyticsService;
    }

    public void AddPerson(T person)
    {
        _people.Add(person);
        
        // Подписка на события человека
        person.PersonActionPerformed += (sender, e) => 
        {
            OnPersonActionLogged(person, $"{e.ActionType}: {e.Description}");
        };

        person.IncomeChanged += (sender, e) =>
        {
            Console.WriteLine($"💰 Изменение дохода: {person.Name} - {e.OldIncome:C} → {e.NewIncome:C}");
        };

        Console.WriteLine($"{person.Name} добавлен(а) в реестр.");
        _analyticsService.TrackPersonCreation(person);
        
        OnPersonAdded(person, "Добавлен в реестр");
        
        if (person is IIdentifiable identifiable)
        {
            Console.WriteLine($"ID объекта: {identifiable.GetId()}, Тип: {identifiable.GetEntityType()}");
        }
    }

    protected virtual void OnPersonAdded(T person, string action)
    {
        PersonAdded?.Invoke(this, new PersonEventArgs(person, action));
    }

    protected virtual void OnPersonActionLogged(T person, string action)
    {
        PersonActionLogged?.Invoke(this, new PersonEventArgs(person, action));
        _analyticsService.TrackInteraction("person_action", person, person);
    }

    public void ValidateAllIds()
    {
        Console.WriteLine("\n=== ПРОВЕРКА ИДЕНТИФИКАТОРОВ ===");
        foreach (var person in _people.OfType<IIdentifiable>())
        {
            bool isValid = person.ValidateId();
            string status = isValid ? "✓ ВАЛИДЕН" : "✗ НЕВАЛИДЕН";
            Console.WriteLine($"{person.GetEntityType()} {person.GetId()}: {status}");
        }
    }

    public void PrintAllIdentifiableInfo()
    {
        Console.WriteLine("\n=== ИНФОРМАЦИЯ ОБ ИДЕНТИФИЦИРУЕМЫХ ОБЪЕКТАХ ===");
        foreach (var identifiable in _people.OfType<IIdentifiable>())
        {
            Console.WriteLine($"{identifiable.GetEntityType()}: {identifiable.GetId()} " +
                            $"(создан: {identifiable.GetCreatedDate():yyyy-MM-dd})");
        }
    }

    public T FindPerson(string name)
    {
        return _people.FirstOrDefault(p => p.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }

    public T FindPerson(string id, bool isIdSearch)
    {
        if (!isIdSearch) return FindPerson(id);
        
        foreach (var person in _people.OfType<IIdentifiable>())
        {
            if (person.GetId() == id)
                return (T)person;
        }
        return null;
    }

    public List<T> GetAllPeople()
    {
        return new List<T>(_people);
    }

    public List<T> GetPeopleByAge(int minAge, int maxAge)
    {
        return _people.Where(p => p.Age >= minAge && p.Age <= maxAge).ToList();
    }

    public void PrintRegistryStats()
    {
        Console.WriteLine($"\n=== СТАТИСТИКА РЕЕСТРА ===");
        Console.WriteLine($"Всего людей: {_people.Count}");
        Console.WriteLine($"Студентов: {_people.OfType<Student>().Count()}");
        Console.WriteLine($"Преподавателей: {_people.OfType<Teacher>().Count()}");
        Console.WriteLine($"Сотрудников: {_people.OfType<Employee>().Count()}");
        Console.WriteLine($"Средний возраст: {_people.Average(p => p.Age):F1} лет");
        Console.WriteLine($"Средний месячный доход: {_people.Average(p => p.MonthlyIncome):C}");
        
        var identifiableCount = _people.OfType<IIdentifiable>().Count();
        Console.WriteLine($"Идентифицируемых объектов: {identifiableCount}/{_people.Count}");
    }

    public void SendBulkNotificationToStudents(string message)
    {
        var students = _people.OfType<Student>().Cast<Person>().ToList();
        _notificationService.SendBulkNotification(message, students);
    }
}

// Глобальный код (точка входа программы)
Console.WriteLine("=== РАСШИРЕННАЯ СИСТЕМА С КОЛЛЕКЦИЯМИ, ДЕЛЕГАТАМИ И СОБЫТИЯМИ ===\n");

// Регистрация зависимостей
DependencyContainer.Register<INotificationService>(new EmailNotificationService());
DependencyContainer.Register<IAnalyticsService>(new AnalyticsService());

var notificationService = DependencyContainer.Resolve<INotificationService>();
var analyticsService = DependencyContainer.Resolve<IAnalyticsService>();

PersonRegistry<Person> registry = new PersonRegistry<Person>(notificationService, analyticsService);

// Подписка на события реестра
registry.PersonAdded += (sender, e) =>
{
    Console.WriteLine($"📝 [РЕЕСТР] {e.Person.Name} - {e.Action}");
};

registry.PersonActionLogged += (sender, e) =>
{
    Console.WriteLine($"📋 [ДЕЙСТВИЕ] {e.Person.Name}: {e.Action}");
};

// Создание объектов
Student student1 = new Student("Артур", 19, "Мужской", "alshin06@gmail.com", "+79526759898", 
    false, "ТИУ", 2, "S2024001", 4.2, "Информационные технологии", true,
    "Компьютерные науки", 45, true, 5000);

Employee employee1 = new Employee("Иван", 35, "Мужской", "ivan@sber.ru", "+79501234567", 
    true, "Сбербанк", 75000, "E1001", 21, "ИТ-отдел", 5, "Менеджер проектов",
    "Петр Сидоров", "9-18", false, 12);

Teacher teacher1 = new Teacher("Мария", 45, "Женский", "maria@university.ru", "+79215555555", 
    true, "История", 13, "T050", "101A", "Доктор наук", 15,
    "Средневековая история", 8, true, 90000);

// Подписка на индивидуальные события
student1.LanguageAdded += (sender, e) =>
{
    Console.WriteLine($"🎯 [ЯЗЫК] {((Person)sender).Name} выучил(а) {e.Language}");
};

employee1.IncomeChanged += (sender, e) =>
{
    Console.WriteLine($"💸 [ДОХОД] {((Person)sender).Name}: {e.OldIncome:C} → {e.NewIncome:C} ({e.Reason})");
};

teacher1.ResearchPublished += (sender, e) =>
{
    Console.WriteLine($"📚 [ИССЛЕДОВАНИЕ] {((Teacher)sender).Name} опубликовал(а) новую работу");
};

// Добавление в реестр
registry.AddPerson(student1);
registry.AddPerson(employee1);
registry.AddPerson(teacher1);

// Демонстрация новых возможностей
Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ НОВЫХ ВОЗМОЖНОСТЕЙ ===");

// Работа с коллекциями Student
student1.AddCourse("Машинное обучение");
student1.AddCourseGrade("Машинное обучение", 4.5);
student1.AddExtracurricularActivity("Шахматный клуб");
student1.PushAssignment("Курсовая работа по ИИ");
student1.AddAcademicGoal("Завершить бакалавриат с отличием");
student1.DisplayAcademicProgress();

// Работа с коллекциями Employee
employee1.AddProjectWithBudget("Автоматизация отчетности", 500000);
employee1.AddTask("Подготовить квартальный отчет");
employee1.AddTask("Встреча с клиентом");
employee1.ScheduleMeeting(DateTime.Now.AddDays(2), "Обсуждение нового проекта");
employee1.RequestVacation(DateTime.Now.AddMonths(1));
employee1.DisplayWorkSummary();

// Работа с коллекциями Teacher
teacher1.AddCourseMaterial("История Средневековья", "Лекция 1: Введение");
teacher1.AddCourseMaterial("История Средневековья", "Презентация: Рыцарство");
teacher1.AddStudentEvaluation(student1, "Отличные успехи в изучении истории");
teacher1.PushPublication("Новые подходы к изучению средневековых документов");
teacher1.AddResearchDeadline(DateTime.Now.AddMonths(2));
teacher1.DisplayTeachingPortfolio();

// Демонстрация работы с базовыми коллекциями Person
Console.WriteLine("\n=== РАБОТА С БАЗОВЫМИ КОЛЛЕКЦИЯМИ ===");
student1.AddSkill("Программирование");
student1.AddSkill("Анализ данных");
student1.AddLanguage("Английский");
student1.AddLanguage("Немецкий");
student1.UpdateContactPreference("SMS", "Disabled");
student1.CompleteTraining("Основы проектного управления");
student1.SendMessage("Привет! Как дела?");
student1.SendMessage("Напоминание о встрече завтра");
student1.DisplaySkills();
student1.DisplayContactPreferences();

// Демонстрация изменения дохода с событием
employee1.UpdateIncome(80000, "Ежегодное повышение");

// Статистика и отчеты
registry.PrintRegistryStats();
registry.ValidateAllIds();
analyticsService.GenerateReport();

// Демонстрация обработки очередей
Console.WriteLine("\n=== ОБРАБОТКА ОЧЕРЕДЕЙ ===");
Console.WriteLine($"Следующее сообщение студента: {student1.GetNextMessage()}");
Console.WriteLine($"Следующая задача сотрудника: {employee1.ProcessNextTask()}");
Console.WriteLine($"Последняя публикация преподавателя: {teacher1.PopPublication()}");

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА ===");
```

Описание задачи:
Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Person:
• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).
• Методы:
o GetInfo(): метод для получения информации о человеке в виде строки.
o SayHello(): метод для вывода приветствия от человека.
Требования к производным классам:
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как
Университет (University) и Курс (Course). Метод SayHello() должен быть
переопределен для добавления информации о курсе при приветствии.
2. Работник (Employee): Должен содержать дополнительные атрибуты, такие
как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть
переопределен для включения информации о компании и зарплате.
3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы
(Experience). Метод SayHello() должен быть переопределен для добавления
информации о преподаваемом предмете при приветствии.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
// ДОБАВЛЯЕМ НОВЫЕ АТРИБУТЫ И МЕТОДЫ В СУЩЕСТВУЮЩИЕ КЛАССЫ
public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string Email { get; set; }
    public string PhoneNumber { get; set; }
    public bool IsMarried { get; set; }

    // Конструктор по умолчанию
    public Person() { }

    // Основной конструктор с использованием свойств (ОБНОВЛЕН)
    public Person(string name, int age, string gender, string email, string phoneNumber, bool isMarried)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Email = email;
        PhoneNumber = phoneNumber;
        IsMarried = isMarried;
    }

    // Конструктор копирования (ОБНОВЛЕН)
    public Person(Person otherPerson)
    {
        Name = otherPerson.Name;
        Age = otherPerson.Age;
        Gender = otherPerson.Gender;
        Email = otherPerson.Email;
        PhoneNumber = otherPerson.PhoneNumber;
        IsMarried = otherPerson.IsMarried;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {Email}, Телефон: {PhoneNumber}, В браке: {(IsMarried ? "Да" : "Нет")}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}");
    }

    // Метод для взаимодействия с другим человеком
    public virtual void InteractWith(Person otherPerson)
    {
        Console.WriteLine($"{Name} знакомится с {otherPerson.Name}");
        SayHello();
        otherPerson.SayHello();
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public virtual void CelebrateBirthday()
    {
        Age++;
        Console.WriteLine($"С Днем Рождения, {Name}! Теперь вам {Age} лет.");
    }

    public string GetContactInfo()
    {
        return $"Контактная информация {Name}: Email: {Email}, Телефон: {PhoneNumber}";
    }
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }
    public List<string> Courses { get; set; }
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string StudentId { get; set; }
    public double AverageGrade { get; set; }

    // Конструктор с использованием свойств (ОБНОВЛЕН)
    public Student(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string university, int course, string studentId, double averageGrade)
        : base(name, age, gender, email, phoneNumber, isMarried) // Вызов конструктора базового класса
    {
        University = university;
        Course = course;
        StudentId = studentId;
        AverageGrade = averageGrade;
        Courses = new List<string>();
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}. Я учусь на {Course} курсе в {University}. Мой средний балл: {AverageGrade:F1}");
    }

    // Взаимодействие студента с преподавателем
    public void AskQuestion(Teacher teacher, string question)
    {
        Console.WriteLine($"{Name} задает вопрос преподавателю {teacher.Name}: '{question}'");
        teacher.AnswerQuestion(this, question);
    }

    // Переопределяем метод взаимодействия для студентов
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Teacher teacher)
        {
            Console.WriteLine($"{Name} обращается к преподавателю {teacher.Name}");
            AskQuestion(teacher, "Можно ли сдать работу позже?");
        }
        else if (otherPerson is Student fellowStudent)
        {
            Console.WriteLine($"{Name} общается с однокурсником {fellowStudent.Name}");
            Console.WriteLine($"{Name}: Привет! Как дела на {fellowStudent.Course} курсе? Твой средний балл {fellowStudent.AverageGrade:F1} - это круто!");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public void Study()
    {
        Console.WriteLine($"{Name} усердно учится в библиотеке. Средний балл может повыситься!");
        // Логика для небольшого увеличения AverageGrade могла бы быть здесь
    }

    public void UpdateGrade(string course, double grade)
    {
        Console.WriteLine($"{Name} получил(а) оценку {grade} по предмету '{course}'.");
        // Простая логика пересчета среднего балла (для примера)
        AverageGrade = (AverageGrade + grade) / 2;
        Console.WriteLine($"Новый средний балл: {AverageGrade:F1}");
    }
}

public class Employee : Person
{
    public string Company { get; set; }
    public decimal Salary { get; set; }
    public string Position { get; set; }
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string EmployeeId { get; set; }
    public int VacationDays { get; set; }

    // Конструктор с использованием свойств (ОБНОВЛЕН)
    public Employee(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string company, decimal salary, string employeeId, int vacationDays, string position = "Сотрудник")
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        Company = company;
        Salary = salary;
        Position = position;
        EmployeeId = employeeId;
        VacationDays = vacationDays;
    }

    public override string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Компания: {Company}, Должность: {Position}, Зарплата: {Salary}, ID: {EmployeeId}, Отпускных дней: {VacationDays}";
    }

    // Метод для работы с другими сотрудниками
    public void WorkWith(Employee colleague)
    {
        Console.WriteLine($"{Name} работает вместе с {colleague.Name} над проектом");
    }

    // Переопределяем метод взаимодействия для сотрудников
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Employee colleague && colleague.Company == Company)
        {
            Console.WriteLine($"{Name} взаимодействует с коллегой {colleague.Name} (ID: {colleague.EmployeeId})");
            WorkWith(colleague);
        }
        else if (otherPerson is Student student)
        {
            Console.WriteLine($"{Name} проводит собеседование со студентом {student.Name} из {student.University}");
            Console.WriteLine($"{Name}: Расскажите о вашем опыте в {student.University}. Ваш средний балл {student.AverageGrade:F1} впечатляет!");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public void Work()
    {
        Console.WriteLine($"{Name} усердно работает на должности '{Position}'. Зарплата {Salary} не начисляется просто так!");
    }

    public void TakeVacation(int days)
    {
        if (days <= VacationDays)
        {
            VacationDays -= days;
            Console.WriteLine($"{Name} уходит в отпуск на {days} дней. Осталось отпускных дней: {VacationDays}.");
        }
        else
        {
            Console.WriteLine($"{Name}, у вас недостаточно отпускных дней. Запрошено: {days}, доступно: {VacationDays}.");
        }
    }
}

public class Teacher : Person
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public List<Student> Students { get; set; }
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string TeacherId { get; set; }
    public string OfficeNumber { get; set; }

    // Конструктор с использованием свойств (ОБНОВЛЕН)
    public Teacher(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string subject, int experience, string teacherId, string officeNumber)
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        Subject = subject;
        Experience = experience;
        TeacherId = teacherId;
        OfficeNumber = officeNumber;
        Students = new List<Student>();
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}. Я преподаю {Subject} уже {Experience} лет. Мой кабинет: {OfficeNumber}");
    }

    // Метод для ответа на вопросы студентов
    public void AnswerQuestion(Student student, string question)
    {
        Console.WriteLine($"{Name} отвечает студенту {student.Name}: 'Хороший вопрос! Давайте разберем его на следующей лекции.'");
    }

    // Метод для добавления студента в список
    public void AddStudent(Student student)
    {
        Students.Add(student);
        Console.WriteLine($"{student.Name} добавлен(а) в список студентов преподавателя {Name}");
    }

    // Метод для проведения лекции
    public void ConductLecture()
    {
        Console.WriteLine($"\n{Name} проводит лекцию по {Subject} (ID преподавателя: {TeacherId})");
        foreach (var student in Students)
        {
            Console.WriteLine($"- {student.Name} (ID: {student.StudentId}) присутствует на лекции");
        }
    }

    // Переопределяем метод взаимодействия для преподавателей
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Student student)
        {
            Console.WriteLine($"{Name} взаимодействует со студентом {student.Name}");
            if (!Students.Contains(student))
            {
                AddStudent(student);
            }
            AnswerQuestion(student, "Общий вопрос по предмету");
        }
        else if (otherPerson is Teacher colleague)
        {
            Console.WriteLine($"{Name} обсуждает методики преподавания с {colleague.Name} (Кабинет: {colleague.OfficeNumber})");
            Console.WriteLine($"{Name}: Как вам новый учебный план по {Subject}?");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public void GradeAssignment(Student student, string course, double grade)
    {
        Console.WriteLine($"Преподаватель {Name} выставил(а) оценку {grade} студенту {student.Name} за курс '{course}'.");
        student.UpdateGrade(course, grade); // Используем новый метод студента
    }

    public void HoldConsultation()
    {
        Console.WriteLine($"Преподаватель {Name} проводит консультацию в кабинете {OfficeNumber}. Приглашаются все студенты курса {Subject}.");
    }
}

// --- СЛОЖНОЕ НАСЛЕДОВАНИЕ: Researcher наследует от Teacher ---
public class Researcher : Teacher
{
    // Новый атрибут
    public string ResearchField { get; set; }

    // Конструктор
    public Researcher(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string subject, int experience, string teacherId, string officeNumber, string researchField)
        : base(name, age, gender, email, phoneNumber, isMarried, subject, experience, teacherId, officeNumber)
    {
        ResearchField = researchField;
    }

    // Переопределение метода
    public override void SayHello()
    {
        base.SayHello(); // Используем метод базового класса (Teacher)
        Console.WriteLine($"А также я провожу исследования в области {ResearchField}.");
    }

    // Новый метод
    public void PublishPaper(string paperTitle)
    {
        Console.WriteLine($"Исследователь {Name} опубликовал научную работу: '{paperTitle}' в области {ResearchField}.");
    }
}

// --- МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (через интерфейс) ---
public interface IManager
{
    string Department { get; set; }
    int TeamSize { get; set; }
    void ConductMeeting();
    void EvaluatePerformance(Employee employee);
}

// Класс, наследующий от Employee и реализующий интерфейс IManager
public class DepartmentHead : Employee, IManager
{
    // Реализация свойств интерфейса
    public string Department { get; set; }
    public int TeamSize { get; set; }

    // Конструктор
    public DepartmentHead(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string company, decimal salary, string employeeId, int vacationDays, string department, int teamSize, string position = "Начальник отдела")
        : base(name, age, gender, email, phoneNumber, isMarried, company, salary, employeeId, vacationDays, position)
    {
        Department = department;
        TeamSize = teamSize;
    }

    // Переопределение метода GetInfo
    public override string GetInfo()
    {
        return base.GetInfo() + $", Отдел: {Department}, Размер команды: {TeamSize}";
    }

    // Реализация методов интерфейса
    public void ConductMeeting()
    {
        Console.WriteLine($"Начальник отдела {Name} проводит собрание отдела '{Department}' с командой из {TeamSize} человек.");
    }

    public void EvaluatePerformance(Employee employee)
    {
        Console.WriteLine($"Начальник отдела {Name} оценивает работу сотрудника {employee.Name}. Вердикт: 'Отлично! Премия!'");
    }

    // Свой метод
    public void MakeStrategicDecision(string decision)
    {
        Console.WriteLine($"Начальник отдела {Name} принимает стратегическое решение для отдела '{Department}': {decision}");
    }
}

// Основной код программы (без класса Program)
Console.WriteLine("=== ДЕМОНСТРАЦИЯ РАБОТЫ РАСШИРЕННЫХ КЛАССОВ ===\n");

// Создаем объекты с использованием ОБНОВЛЕННЫХ конструкторов
Student student1 = new Student("Артур", 19, "Мужской", "alshin06@gmail.com", "+79526759898", false, "ТИУ", 2, "S2024001", 4.2);
Student student2 = new Student("Анна", 20, "Женский", "anna@university.ru", "+79098765432", false, "ТИУ", 3, "S2023005", 4.8);

Employee employee1 = new Employee("Иван", 35, "Мужской", "ivan@sber.ru", "+79501234567", true, "Сбербанк", 75000, "E1001", 21, "Менеджер");
Employee employee2 = new Employee("Елена", 28, "Женский", "elena@sber.ru", "+79507654321", false, "Сбербанк", 65000, "E1002", 18, "Аналитик");

Teacher teacher1 = new Teacher("Мария", 45, "Женский", "maria@university.ru", "+79215555555", true, "История", 13, "T050", "101A");
Teacher teacher2 = new Teacher("Сергей", 50, "Мужской", "sergey@university.ru", "+79216666666", true, "Информатика", 20, "T025", "205B");

// --- ДЕМОНСТРАЦИЯ СЛОЖНОГО НАСЛЕДОВАНИЯ ---
Console.WriteLine("=== ДЕМОНСТРАЦИЯ СЛОЖНОГО НАСЛЕДОВАНИЯ (Researcher) ===");
Researcher researcher1 = new Researcher("Алексей", 55, "Мужской", "alexey@science.ru", "+79217777777", true, "Физика", 25, "T010", "301", "Квантовая механика");
researcher1.SayHello(); // Вызов переопределенного метода
researcher1.PublishPaper("Квантовые вычисления для начинающих");
researcher1.ConductLecture(); // Метод, унаследованный от Teacher
Console.WriteLine();

// --- ДЕМОНСТРАЦИЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ (через интерфейс) ---
Console.WriteLine("=== ДЕМОНСТРАЦИЯ 'МНОЖЕСТВЕННОГО' НАСЛЕДОВАНИЯ (DepartmentHead) ===");
DepartmentHead head1 = new DepartmentHead("Ольга", 40, "Женский", "olga@sber.ru", "+79509998877", true, "Сбербанк", 150000, "E9001", 28, "ИТ-Разработка", 15);
head1.SayHello();
Console.WriteLine(head1.GetInfo()); // Вызов переопределенного метода
head1.ConductMeeting();    // Метод из интерфейса IManager
head1.EvaluatePerformance(employee2); // Метод из интерфейса IManager
head1.Work();              // Метод, унаследованный от Employee
head1.TakeVacation(5);     // Метод, унаследованный от Employee
Console.WriteLine();

// Создаем массив для демонстрации полиморфизма (ДОБАВИЛИ НОВЫЕ ОБЪЕКТЫ)
Person[] people = new Person[8];
people[0] = student1;
people[1] = student2;
people[2] = employee1;
people[3] = employee2;
people[4] = teacher1;
people[5] = teacher2;
people[6] = researcher1; // Researcher является и Person, и Teacher
people[7] = head1;       // DepartmentHead является и Person, и Employee

Console.WriteLine("=== ИНФОРМАЦИЯ О ВСЕХ ЛЮДЯХ (ПОЛИМОРФИЗМ) ===");
foreach (var person in people)
{
    person.SayHello(); // Полиморфный вызов
    Console.WriteLine(person.GetInfo()); // Полиморфный вызов
    Console.WriteLine();
}

Console.WriteLine("=== ВЗАИМОДЕЙСТВИЕ МЕЖДУ ОБЪЕКТАМИ ===\n");

// Демонстрация различных типов взаимодействия
student1.InteractWith(teacher1);
Console.WriteLine();

student1.InteractWith(student2);
Console.WriteLine();

employee1.InteractWith(employee2);
Console.WriteLine();

teacher1.InteractWith(student2);
Console.WriteLine();

employee1.InteractWith(student1);
Console.WriteLine();

teacher1.InteractWith(teacher2);
Console.WriteLine();

// Демонстрация работы НОВЫХ методов конкретных классов
Console.WriteLine("=== СПЕЦИФИЧЕСКИЕ ДЕЙСТВИЯ (НОВЫЕ МЕТОДЫ) ===");
teacher1.AddStudent(student1);
teacher1.AddStudent(student2);
teacher1.ConductLecture();
teacher1.GradeAssignment(student1, "История Древнего Мира", 4.5); // Новый метод
teacher1.HoldConsultation(); // Новый метод
Console.WriteLine();

student1.Study(); // Новый метод
student1.UpdateGrade("Программирование", 5.0); // Новый метод
Console.WriteLine();

employee1.Work(); // Новый метод
employee1.TakeVacation(5); // Новый метод
Console.WriteLine();

Console.WriteLine("\n=== ВОПРОС СТУДЕНТА ПРЕПОДАВАТЕЛЮ ===");
student1.AskQuestion(teacher1, "Как правильно подготовиться к экзамену?");
```

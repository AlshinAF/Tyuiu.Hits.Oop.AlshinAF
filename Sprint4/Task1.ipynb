<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также применить явнаю реализацию интерфейса.

````
// Интерфейс для животных
public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
}

// Базовый абстрактный класс Animal
public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    
    protected Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }
    
    // Виртуальный метод с реализацией по умолчанию
    public virtual string GetInfo()
    {
        return $"{GetType().Name} {Name}, возраст: {Age} лет, вес: {Weight} кг";
    }
}

// Класс Dog с неявной реализацией интерфейса
public class Dog : Animal, IAnimal
{
    public string Breed { get; set; }
    
    public Dog(string name, int age, double weight, string breed) 
        : base(name, age, weight)
    {
        Breed = breed;
    }
    
    // Неявная реализация метода MakeSound
    public void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    // Неявная реализация метода Move
    public void Move()
    {
        Console.WriteLine($"{Name} бегает на четырех лапах");
    }
    
    // Неявная реализация метода GetInfo
    public new string GetInfo()
    {
        return base.GetInfo() + $", порода: {Breed}";
    }
}

// Класс Cat с явной реализацией интерфейса
public class Cat : Animal, IAnimal
{
    public string Color { get; set; }
    
    public Cat(string name, int age, double weight, string color) 
        : base(name, age, weight)
    {
        Color = color;
    }
    
    // Явная реализация метода MakeSound
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    // Явная реализация метода Move
    void IAnimal.Move()
    {
        Console.WriteLine($"{Name} грациозно ходит");
    }
    
    // Явная реализация метода GetInfo
    string IAnimal.GetInfo()
    {
        return $"{GetType().Name} {Name} ({Color}), возраст: {Age} лет, вес: {Weight} кг";
    }
    
    // Собственный метод класса
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
    
    // Переопределение базового метода
    public override string GetInfo()
    {
        return $"{GetType().Name} {Name} ({Color}), возраст: {Age} лет, вес: {Weight} кг";
    }
}

// Класс Lion с явной реализацией интерфейса
public class Lion : Animal, IAnimal
{
    public string ManeColor { get; set; }
    
    public Lion(string name, int age, double weight, string maneColor) 
        : base(name, age, weight)
    {
        ManeColor = maneColor;
    }
    
    // Явная реализация всех методов интерфейса
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"{Name} рычит: Ррррррр!");
    }
    
    void IAnimal.Move()
    {
        Console.WriteLine($"{Name} величественно шествует");
    }
    
    string IAnimal.GetInfo()
    {
        return $"{GetType().Name} {Name}, цвет гривы: {ManeColor}, возраст: {Age} лет, вес: {Weight} кг";
    }
    
    // Собственный метод класса
    public void Hunt()
    {
        Console.WriteLine($"{Name} отправляется на охоту!");
    }
    
    // Переопределение базового метода
    public override string GetInfo()
    {
        return $"{GetType().Name} {Name}, цвет гривы: {ManeColor}, возраст: {Age} лет, вес: {Weight} кг";
    }
}

// Класс Bird для разнообразия
public class Bird : Animal, IAnimal
{
    public double Wingspan { get; set; }
    
    public Bird(string name, int age, double weight, double wingspan) 
        : base(name, age, weight)
    {
        Wingspan = wingspan;
    }
    
    // Неявная реализация
    public void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public void Move()
    {
        Console.WriteLine($"{Name} летает в небе");
    }
    
    public new string GetInfo()
    {
        return base.GetInfo() + $", размах крыльев: {Wingspan} см";
    }
}
// Создаем животных
Dog umka = new Dog("Умка", 3, 25.5, "Хаски");
Cat petrovich = new Cat("Петрович", 5, 4.2, "Рыжий");
Lion simba = new Lion("Симба", 7, 190.0, "Золотистая");
Bird kesha = new Bird("Кеша", 2, 0.3, 15.5);

Console.WriteLine("=== Демонстрация работы с животными ===\n");

// Работа с собакой (неявная реализация)
Console.WriteLine("Собака (неявная реализация):");
Console.WriteLine(umka.GetInfo());
umka.MakeSound();
umka.Move();
Console.WriteLine();

// Работа с котом (явная реализация - методы доступны только через интерфейс)
Console.WriteLine("Кот (явная реализация):");
Console.WriteLine(petrovich.GetInfo()); // Базовый метод
IAnimal petrovichAnimal = petrovich;
Console.WriteLine(petrovichAnimal.GetInfo()); // Метод интерфейса
petrovichAnimal.MakeSound();
petrovichAnimal.Move();
petrovich.Purr(); // Собственный метод доступен напрямую
Console.WriteLine();

// Работа со львом (явная реализация - методы доступны только через интерфейс)
Console.WriteLine("Лев (явная реализация):");
Console.WriteLine(simba.GetInfo()); // Базовый метод
IAnimal simbaAnimal = simba;
Console.WriteLine(simbaAnimal.GetInfo()); // Метод интерфейса
simbaAnimal.MakeSound();
simbaAnimal.Move();
simba.Hunt(); // Собственный метод доступен напрямую
Console.WriteLine();

// Работа с птицей
Console.WriteLine("Птица (неявная реализация):");
Console.WriteLine(kesha.GetInfo());
kesha.MakeSound();
kesha.Move();
Console.WriteLine();

// Демонстрация полиморфизма
Console.WriteLine("=== Полиморфизм через интерфейс ===");
List<IAnimal> animals = new List<IAnimal> { umka, petrovich, simba, kesha };

foreach (var animal in animals)
{
    Console.WriteLine("---");
    Console.WriteLine(animal.GetInfo());
    animal.MakeSound();
    animal.Move();
    
    // Проверка типа для вызова специфических методов
    if (animal is Cat cat)
    {
        cat.Purr();
    }
    else if (animal is Lion lion)
    {
        lion.Hunt();
    }
}
````

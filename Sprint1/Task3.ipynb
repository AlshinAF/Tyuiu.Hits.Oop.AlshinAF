Описание задачи:
Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Person:
• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).
• Методы:
o GetInfo(): метод для получения информации о человеке в виде строки.
o SayHello(): метод для вывода приветствия от человека.
Требования к производным классам:
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как
Университет (University) и Курс (Course). Метод SayHello() должен быть
переопределен для добавления информации о курсе при приветствии.
2. Работник (Employee): Должен содержать дополнительные атрибуты, такие
как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть
переопределен для включения информации о компании и зарплате.
3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы
(Experience). Метод SayHello() должен быть переопределен для добавления
информации о преподаваемом предмете при приветствии.


#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

```
public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }

    // Конструктор по умолчанию
    public Person() { }

    // Основной конструктор с использованием свойств
    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    // Конструктор копирования
    public Person(Person otherPerson)
    {
        Name = otherPerson.Name;
        Age = otherPerson.Age;
        Gender = otherPerson.Gender;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}");
    }

    // Метод для взаимодействия с другим человеком
    public virtual void InteractWith(Person otherPerson)
    {
        Console.WriteLine($"{Name} знакомится с {otherPerson.Name}");
        SayHello();
        otherPerson.SayHello();
    }
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }
    public List<string> Courses { get; set; }

    // Конструктор с использованием свойств
    public Student(string name, int age, string gender, string university, int course)
    {
        Name = name;
        Age = age;
        Gender = gender;
        University = university;
        Course = course;
        Courses = new List<string>();
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}. Я учусь на {Course} курсе в {University}");
    }

    // Взаимодействие студента с преподавателем
    public void AskQuestion(Teacher teacher, string question)
    {
        Console.WriteLine($"{Name} задает вопрос преподавателю {teacher.Name}: '{question}'");
        teacher.AnswerQuestion(this, question);
    }

    // Переопределяем метод взаимодействия для студентов
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Teacher teacher)
        {
            Console.WriteLine($"{Name} обращается к преподавателю {teacher.Name}");
            AskQuestion(teacher, "Можно ли сдать работу позже?");
        }
        else if (otherPerson is Student fellowStudent)
        {
            Console.WriteLine($"{Name} общается с однокурсником {fellowStudent.Name}");
            Console.WriteLine($"{Name}: Привет! Как дела на {fellowStudent.Course} курсе?");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }
}

public class Employee : Person
{
    public string Company { get; set; }
    public decimal Salary { get; set; }
    public string Position { get; set; }

    // Конструктор с использованием свойств
    public Employee(string name, int age, string gender, string company, decimal salary, string position = "Сотрудник")
    {
        Name = name;
        Age = age;
        Gender = gender;
        Company = company;
        Salary = salary;
        Position = position;
    }

    public override string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Компания: {Company}, Должность: {Position}, Зарплата: {Salary}";
    }

    // Метод для работы с другими сотрудниками
    public void WorkWith(Employee colleague)
    {
        Console.WriteLine($"{Name} работает вместе с {colleague.Name} над проектом");
    }

    // Переопределяем метод взаимодействия для сотрудников
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Employee colleague && colleague.Company == Company)
        {
            Console.WriteLine($"{Name} взаимодействует с коллегой {colleague.Name}");
            WorkWith(colleague);
        }
        else if (otherPerson is Student student)
        {
            Console.WriteLine($"{Name} проводит собеседование со студентом {student.Name}");
            Console.WriteLine($"{Name}: Расскажите о вашем опыте в {student.University}");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }
}

public class Teacher : Person
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public List<Student> Students { get; set; }

    // Конструктор с использованием свойств
    public Teacher(string name, int age, string gender, string subject, int experience)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Subject = subject;
        Experience = experience;
        Students = new List<Student>();
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}. Я преподаю {Subject} уже {Experience} лет");
    }

    // Метод для ответа на вопросы студентов
    public void AnswerQuestion(Student student, string question)
    {
        Console.WriteLine($"{Name} отвечает студенту {student.Name}: 'Хороший вопрос! Давайте разберем его на следующей лекции.'");
    }

    // Метод для добавления студента в список
    public void AddStudent(Student student)
    {
        Students.Add(student);
        Console.WriteLine($"{student.Name} добавлен(а) в список студентов преподавателя {Name}");
    }

    // Метод для проведения лекции
    public void ConductLecture()
    {
        Console.WriteLine($"\n{Name} проводит лекцию по {Subject}");
        foreach (var student in Students)
        {
            Console.WriteLine($"- {student.Name} присутствует на лекции");
        }
    }

    // Переопределяем метод взаимодействия для преподавателей
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Student student)
        {
            Console.WriteLine($"{Name} взаимодействует со студентом {student.Name}");
            if (!Students.Contains(student))
            {
                AddStudent(student);
            }
            AnswerQuestion(student, "Общий вопрос по предмету");
        }
        else if (otherPerson is Teacher colleague)
        {
            Console.WriteLine($"{Name} обсуждает методики преподавания с {colleague.Name}");
            Console.WriteLine($"{Name}: Как вам новый учебный план по {Subject}?");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }
}

// Основной код программы (без класса Program)
Console.WriteLine("=== ДЕМОНСТРАЦИЯ РАБОТЫ КЛАССОВ ===\n");

// Создаем объекты с использованием конструкторов
Student student1 = new Student("Артур", 19, "Мужской", "ТИУ", 2);
Student student2 = new Student("Анна", 20, "Женский", "ТИУ", 3);

Employee employee1 = new Employee("Иван", 35, "Мужской", "Сбербанк", 75000, "Менеджер");
Employee employee2 = new Employee("Елена", 28, "Женский", "Сбербанк", 65000, "Аналитик");

Teacher teacher1 = new Teacher("Мария", 45, "Женский", "История", 13);
Teacher teacher2 = new Teacher("Сергей", 50, "Мужской", "Информатика", 20);

// Создаем массив для демонстрации полиморфизма
Person[] people = new Person[6];
people[0] = student1;
people[1] = student2;
people[2] = employee1;
people[3] = employee2;
people[4] = teacher1;
people[5] = teacher2;

Console.WriteLine("=== ИНФОРМАЦИЯ О ВСЕХ ЛЮДЯХ ===");
foreach (var person in people)
{
    person.SayHello();
    Console.WriteLine(person.GetInfo());
    Console.WriteLine();
}

Console.WriteLine("=== ВЗАИМОДЕЙСТВИЕ МЕЖДУ ОБЪЕКТАМИ ===\n");

// Демонстрация различных типов взаимодействия
student1.InteractWith(teacher1);
Console.WriteLine();

student1.InteractWith(student2);
Console.WriteLine();

employee1.InteractWith(employee2);
Console.WriteLine();

teacher1.InteractWith(student2);
Console.WriteLine();

employee1.InteractWith(student1);
Console.WriteLine();

teacher1.InteractWith(teacher2);
Console.WriteLine();

// Демонстрация работы методов конкретных классов
Console.WriteLine("=== СПЕЦИФИЧЕСКИЕ ДЕЙСТВИЯ ===");
teacher1.AddStudent(student1);
teacher1.AddStudent(student2);
teacher1.ConductLecture();

Console.WriteLine("\n=== ВОПРОС СТУДЕНТА ПРЕПОДАВАТЕЛЮ ===");
student1.AskQuestion(teacher1, "Как правильно подготовиться к экзамену?");
```

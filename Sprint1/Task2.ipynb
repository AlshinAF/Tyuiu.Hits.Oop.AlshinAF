### Создание экземпляров классов (объектов)

В C#, экземпляр класса создается с помощью оператора `new`, который выделяет память для нового объекта и вызывает конструктор класса для инициализации этого объекта. Конструктор — это специальный метод в классе, который имеет то же имя, что и класс, и используется для установки начального состояния объекта.

#### Пример создания экземпляров классов Vehicle и Car

```csharp
Vehicle myVehicle = new Vehicle();
Car myCar = new Car();
```

В этом примере создаются два объекта: один для класса `Vehicle` и другой для класса `Car`. Конструкторы этих классов будут автоматически вызваны при создании объектов.

### Взаимодействие объектов между Собой

Взаимодействие объектов между собой означает, что один объект может вызывать методы другого объекта или обращаться к его свойствам (если они доступны). Это позволяет объектам обмениваться информацией и координировать свои действия.

#### Пример взаимодействия объектов

```csharp
myVehicle.Color = "Red";
myVehicle.Speed = 60;

myCar.Color = "Blue";
myCar.FuelLevel = 100;
myCar.Refuel(50); // Увеличиваем уровень топлива на 50
```

В этом примере мы устанавливаем свойства `Color` и `Speed` для объекта `myVehicle`, а также свойства `Color` и `FuelLevel` для объекта `myCar`. Затем мы вызываем метод `Refuel` у объекта `myCar`, чтобы увеличить его уровень топлива.

### Закрепление правил использования модификаторов доступа

Модификаторы доступа определяют уровень доступности членов класса из других частей кода. Важно правильно использовать модификаторы доступа для обеспечения безопасности и инкапсуляции данных.

#### Примеры модификаторов доступа в классах Vehicle и Car

```csharp
public class Vehicle
{
    public string Color { get; set; } // public свойство
    private int speed; // private поле

    protected void SetSpeed(int value) // protected метод
    {
        speed = value;
    }
}

public class Car : Vehicle
{
    private int fuelLevel; // private поле

    public void Refuel(int amount) // public метод
    {
        fuelLevel += amount;
    }

    protected void CheckSpeedLimit() // protected метод
    {
        if (speed > 120)
        {
            Console.WriteLine("Превышен скоростной режим!");
        }
    }
}
```

В этом примере:
- Свойство `Color` в классе `Vehicle` объявлено как `public`, что позволяет доступ к нему из любого места в коде.
- Поле `speed` в классе `Vehicle` объявлено как `private`, что ограничивает доступ к нему только внутри класса `Vehicle`.
- Метод `SetSpeed` в классе `Vehicle` объявлен как `protected`, что позволяет доступ к нему из класса `Vehicle` и его производных классов.
- Поле `fuelLevel` в классе `Car` объявлено как `private`, ограничивая доступ к нему только внутри класса `Car`.
- Методы `Refuel` и `CheckSpeedLimit` в классе `Car` демонстрируют использование модификаторов доступа для контроля над тем, как объекты могут взаимодействовать друг с другом.

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса с реализацией взаимодействия между объектами.

````
// Базовый класс Animal
public class Animal
{
    // Статическое поле - общее для всех животных
    private static int animalCount = 0;
    
    // Обычные поля с инкапсуляцией через свойства
    private string name;
    private int age;
    private double weight;
    
    // Конструктор
    public Animal(string name, int age, double weight)
    {
        this.name = name;
        this.age = age;
        this.weight = weight;
        animalCount++; // Увеличиваем счетчик при создании каждого животного
    }
    
    // Свойства с геттерами и сеттерами
    public string Name
    {
        get { return name; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                name = value;
        }
    }
    
    public int Age
    {
        get { return age; }
        set 
        { 
            if (value >= 0)
                age = value;
        }
    }
    
    public double Weight
    {
        get { return weight; }
        set 
        { 
            if (value > 0)
                weight = value;
        }
    }
    
    // Статическое свойство только для чтения
    public static int AnimalCount
    {
        get { return animalCount; }
    }
    
    // Виртуальный метод - может быть переопределен в производных классах
    public virtual void MakeSound()
    {
        Console.WriteLine($"{name} издает звук");
    }
    
    // Обычный метод
    public void Eat(string food)
    {
        Console.WriteLine($"{name} ест {food}");
        weight += 0.1; // При еде вес немного увеличивается
    }
    
    // Статический метод
    public static void DisplayAnimalCount()
    {
        Console.WriteLine($"Всего создано животных: {animalCount}");
    }
    
    // Защищенный метод - доступен только в этом классе и производных
    protected void Sleep()
    {
        Console.WriteLine($"{name} спит");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    private string breed;
    private bool isTrained;
    
    public Dog(string name, int age, double weight, string breed, bool isTrained) 
        : base(name, age, weight)
    {
        this.breed = breed;
        this.isTrained = isTrained;
    }
    
    // Свойства
    public string Breed
    {
        get { return breed; }
        set { breed = value; }
    }
    
    public bool IsTrained
    {
        get { return isTrained; }
        set { isTrained = value; }
    }
    
    // Переопределение виртуального метода
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    // Специфический метод для собаки
    public void Fetch(string item)
    {
        Console.WriteLine($"{Name} приносит {item}");
        if (isTrained)
        {
            Console.WriteLine($"{Name} хорошо выполняет команду!");
        }
        else
        {
            Console.WriteLine($"{Name} еще учится...");
        }
    }
    
    // Использование защищенного метода базового класса
    public void Rest()
    {
        Sleep(); // Вызываем защищенный метод базового класса
    }
}

// Производный класс Cat
public class Cat : Animal
{
    private string color;
    private bool isIndoor;
    
    public Cat(string name, int age, double weight, string color, bool isIndoor) 
        : base(name, age, weight)
    {
        this.color = color;
        this.isIndoor = isIndoor;
    }
    
    // Свойства
    public string Color
    {
        get { return color; }
        set { color = value; }
    }
    
    public bool IsIndoor
    {
        get { return isIndoor; }
        set { isIndoor = value; }
    }
    
    // Переопределение виртуального метода
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    // Специфический метод для кошки
    public void PlayWithToy(string toy)
    {
        Console.WriteLine($"{Name} играет с {toy}");
        if (!isIndoor)
        {
            Console.WriteLine($"{Name} выходит на улицу поиграть");
        }
    }
    
    // Использование защищенного метода базового класса
    public void TakeNap()
    {
        Sleep(); // Вызываем защищенный метод базового класса
        Console.WriteLine($"{Name} сладко спит на подушке");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    private double wingspan;
    private bool canFly;
    
    public Bird(string name, int age, double weight, double wingspan, bool canFly) 
        : base(name, age, weight)
    {
        this.wingspan = wingspan;
        this.canFly = canFly;
    }
    
    // Свойства
    public double Wingspan
    {
        get { return wingspan; }
        set { wingspan = value; }
    }
    
    public bool CanFly
    {
        get { return canFly; }
        set { canFly = value; }
    }
    
    // Переопределение виртуального метода
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    // Специфический метод для птицы
    public void Fly()
    {
        if (canFly)
        {
            Console.WriteLine($"{Name} летает с размахом крыльев {wingspan} см");
        }
        else
        {
            Console.WriteLine($"{Name} не умеет летать");
        }
    }
}

// Демонстрация взаимодействия объектов
public class Zoo
{
    public void DemonstrateInteraction()
    {
        // Создаем экземпляры животных
        Dog myDog = new Dog("Мухтар", 3, 15.5, "Овчарка", true);
        Cat myCat = new Cat("Пушок", 2, 4.2, "Рыжий", true);
        Bird myBird = new Bird("Кеша", 1, 0.3, 25.0, true);
        
        Console.WriteLine("=== ДЕМОНСТРАЦИЯ ВЗАИМОДЕЙСТВИЯ ОБЪЕКТОВ ===\n");
        
        // Взаимодействие с объектом Dog
        Console.WriteLine("1. Взаимодействие с собакой:");
        myDog.MakeSound();
        myDog.Eat("кость");
        myDog.Fetch("мяч");
        myDog.Rest();
        Console.WriteLine($"Вес собаки: {myDog.Weight} кг\n");
        
        // Взаимодействие с объектом Cat
        Console.WriteLine("2. Взаимодействие с кошкой:");
        myCat.MakeSound();
        myCat.Eat("рыбу");
        myCat.PlayWithToy("клубок");
        myCat.TakeNap();
        Console.WriteLine($"Цвет кошки: {myCat.Color}\n");
        
        // Взаимодействие с объектом Bird
        Console.WriteLine("3. Взаимодействие с птицей:");
        myBird.MakeSound();
        myBird.Eat("зерно");
        myBird.Fly();
        Console.WriteLine($"Размах крыльев: {myBird.Wingspan} см\n");
        
        // Демонстрация статических членов
        Console.WriteLine("4. Статические члены класса:");
        Animal.DisplayAnimalCount();
        Console.WriteLine($"Всего животных: {Animal.AnimalCount}");
        
        // Взаимодействие между разными объектами
        Console.WriteLine("\n5. Взаимодействие между разными животными:");
        Console.WriteLine($"{myDog.Name} видит {myCat.Name} и начинает гавкать");
        myDog.MakeSound();
        Console.WriteLine($"{myCat.Name} в ответ шипит");
        Console.WriteLine($"{myBird.Name} наблюдает за ними сверху");
        myBird.MakeSound();
    }
}

// Создаем экземпляр Zoo и демонстрируем взаимодействие
Zoo zoo = new Zoo();
zoo.DemonstrateInteraction();
````

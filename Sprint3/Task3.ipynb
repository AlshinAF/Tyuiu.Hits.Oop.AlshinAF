```

// Дженерик-класс для управления коллекциями объектов
public class PersonRegistry<T> where T : Person
{
    private List<T> _people = new List<T>();

    // Добавление человека в реестр
    public void AddPerson(T person)
    {
        _people.Add(person);
        Console.WriteLine($"{person.Name} добавлен(а) в реестр.");
    }

    // Поиск по имени (перегрузка метода)
    public T FindPerson(string name)
    {
        return _people.FirstOrDefault(p => p.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }

    // Поиск по идентификатору (перегрузка метода) - демонстрация перегрузки
    public T FindPerson(string id, bool isIdSearch)
    {
        if (!isIdSearch) return FindPerson(id);
        
        foreach (var person in _people)
        {
            if (person is Student student && student.StudentId == id)
                return (T)(Person)student;
            if (person is Teacher teacher && teacher.TeacherId == id)
                return (T)(Person)teacher;
            if (person is Employee employee && employee.EmployeeId == id)
                return (T)(Person)employee;
        }
        return null;
    }

    // Получение всех людей в реестре
    public List<T> GetAllPeople()
    {
        return new List<T>(_people);
    }

    // Получение людей по возрасту (диапазон)
    public List<T> GetPeopleByAge(int minAge, int maxAge)
    {
        return _people.Where(p => p.Age >= minAge && p.Age <= maxAge).ToList();
    }

    // Статистика по реестру
    public void PrintRegistryStats()
    {
        Console.WriteLine($"\n=== СТАТИСТИКА РЕЕСТРА ===");
        Console.WriteLine($"Всего людей: {_people.Count}");
        Console.WriteLine($"Студентов: {_people.OfType<Student>().Count()}");
        Console.WriteLine($"Преподавателей: {_people.OfType<Teacher>().Count()}");
        Console.WriteLine($"Сотрудников: {_people.OfType<Employee>().Count()}");
        Console.WriteLine($"Средний возраст: {_people.Average(p => p.Age):F1} лет");
    }
}

public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }
    
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string Email { get; set; }
    public string PhoneNumber { get; set; }
    public bool IsMarried { get; set; }
    public string Nationality { get; set; } // Новый атрибут
    public DateTime RegistrationDate { get; set; } // Новый атрибут

    // Конструктор по умолчанию
    public Person() 
    {
        RegistrationDate = DateTime.Now;
    }

    // Основной конструктор с использованием свойств (ОБНОВЛЕН)
    public Person(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string nationality = "Не указана")
    {
        Name = name;
        Age = age;
        Gender = gender;
        Email = email;
        PhoneNumber = phoneNumber;
        IsMarried = isMarried;
        Nationality = nationality;
        RegistrationDate = DateTime.Now;
    }

    // Конструктор копирования (ОБНОВЛЕН)
    public Person(Person otherPerson)
    {
        Name = otherPerson.Name;
        Age = otherPerson.Age;
        Gender = otherPerson.Gender;
        Email = otherPerson.Email;
        PhoneNumber = otherPerson.PhoneNumber;
        IsMarried = otherPerson.IsMarried;
        Nationality = otherPerson.Nationality;
        RegistrationDate = otherPerson.RegistrationDate;
    }

    // Переопределяемый метод (полиморфизм с перекрытием)
    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {Email}, Телефон: {PhoneNumber}, В браке: {(IsMarried ? "Да" : "Нет")}, Национальность: {Nationality}";
    }

    // Перегруженные методы SayHello (полиморфизм с перегрузкой)
    public virtual void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}");
    }

    public virtual void SayHello(string greetingType)
    {
        switch (greetingType.ToLower())
        {
            case "formal":
                Console.WriteLine($"Здравствуйте, меня зовут {Name}. Рад(а) познакомиться!");
                break;
            case "friendly":
                Console.WriteLine($"Приветик! Я {Name} :)");
                break;
            case "professional":
                Console.WriteLine($"Добрый день. Мое имя {Name}.");
                break;
            default:
                SayHello();
                break;
        }
    }

    // Перегруженный метод для взаимодействия
    public virtual void InteractWith(Person otherPerson)
    {
        Console.WriteLine($"{Name} знакомится с {otherPerson.Name}");
        SayHello();
        otherPerson.SayHello();
    }

    public virtual void InteractWith(Person otherPerson, string context)
    {
        Console.WriteLine($"{Name} взаимодействует с {otherPerson.Name} в контексте: {context}");
        SayHello("formal");
        otherPerson.SayHello("formal");
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public virtual void CelebrateBirthday()
    {
        Age++;
        Console.WriteLine($"С Днем Рождения, {Name}! Теперь вам {Age} лет.");
    }

    public string GetContactInfo()
    {
        return $"Контактная информация {Name}: Email: {Email}, Телефон: {PhoneNumber}";
    }

    // Новый метод с перегрузкой
    public virtual void UpdateProfile(string email, string phone)
    {
        Email = email;
        PhoneNumber = phone;
        Console.WriteLine($"{Name}: Контактная информация обновлена.");
    }

    public virtual void UpdateProfile(string email, string phone, string nationality)
    {
        Email = email;
        PhoneNumber = phone;
        Nationality = nationality;
        Console.WriteLine($"{Name}: Профиль полностью обновлен.");
    }

    // Новый метод для проверки совершеннолетия
    public bool IsAdult()
    {
        return Age >= 18;
    }
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }
    public List<string> Courses { get; set; }
    
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string StudentId { get; set; }
    public double AverageGrade { get; set; }
    public string Faculty { get; set; } // Новый атрибут
    public bool HasScholarship { get; set; } // Новый атрибут

    // Конструктор с использованием свойств (ОБНОВЛЕН)
    public Student(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string university, int course, string studentId, double averageGrade, string faculty = "Не указан", bool hasScholarship = false)
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        University = university;
        Course = course;
        StudentId = studentId;
        AverageGrade = averageGrade;
        Faculty = faculty;
        HasScholarship = hasScholarship;
        Courses = new List<string>();
    }

    // Переопределение метода (полиморфизм с перекрытием)
    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}. Я учусь на {Course} курсе в {University}. Мой средний балл: {AverageGrade:F1}");
    }

    // Перегрузка метода (полиморфизм с перегрузкой)
    public override void SayHello(string greetingType)
    {
        if (greetingType.ToLower() == "student")
        {
            Console.WriteLine($"Йоу! Я {Name} с {Faculty} факультета. Давай учиться вместе!");
        }
        else
        {
            base.SayHello(greetingType);
        }
    }

    // Переопределение метода GetInfo
    public override string GetInfo()
    {
        return base.GetInfo() + $", Университет: {University}, Факультет: {Faculty}, Курс: {Course}, ID: {StudentId}, Средний балл: {AverageGrade:F1}, Стипендия: {(HasScholarship ? "Да" : "Нет")}";
    }

    // Взаимодействие студента с преподавателем
    public void AskQuestion(Teacher teacher, string question)
    {
        Console.WriteLine($"{Name} задает вопрос преподавателю {teacher.Name}: '{question}'");
        teacher.AnswerQuestion(this, question);
    }

    // Переопределяем метод взаимодействия для студентов
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Teacher teacher)
        {
            Console.WriteLine($"{Name} обращается к преподавателю {teacher.Name}");
            AskQuestion(teacher, "Можно ли сдать работу позже?");
        }
        else if (otherPerson is Student fellowStudent)
        {
            Console.WriteLine($"{Name} общается с однокурсником {fellowStudent.Name}");
            Console.WriteLine($"{Name}: Привет! Как дела на {fellowStudent.Course} курсе? Твой средний балл {fellowStudent.AverageGrade:F1} - это круто!");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public void Study()
    {
        Console.WriteLine($"{Name} усердно учится в библиотеке. Средний балл может повыситься!");
    }

    public void Study(string subject) // Перегрузка метода Study
    {
        Console.WriteLine($"{Name} усердно изучает {subject} в библиотеке.");
        if (!Courses.Contains(subject))
        {
            Courses.Add(subject);
        }
    }

    public void UpdateGrade(string course, double grade)
    {
        Console.WriteLine($"{Name} получил(а) оценку {grade} по предмету '{course}'.");
        AverageGrade = (AverageGrade + grade) / 2;
        Console.WriteLine($"Новый средний балл: {AverageGrade:F1}");
    }

    // Новый метод для применения на стипендию
    public void ApplyForScholarship()
    {
        if (AverageGrade >= 4.5 && !HasScholarship)
        {
            HasScholarship = true;
            Console.WriteLine($"{Name} получил(а) академическую стипендию! Поздравляем!");
        }
        else if (HasScholarship)
        {
            Console.WriteLine($"{Name} уже получает стипендию.");
        }
        else
        {
            Console.WriteLine($"{Name} не соответствует критериям для стипендии. Средний балл должен быть не менее 4.5.");
        }
    }
}

public class Employee : Person
{
    public string Company { get; set; }
    public decimal Salary { get; set; }
    public string Position { get; set; }
    
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string EmployeeId { get; set; }
    public int VacationDays { get; set; }
    public string Department { get; set; } // Новый атрибут
    public int YearsInCompany { get; set; } // Новый атрибут

    // Конструктор с использованием свойств (ОБНОВЛЕН)
    public Employee(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string company, decimal salary, string employeeId, int vacationDays, string department = "Не указан", int yearsInCompany = 0, string position = "Сотрудник")
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        Company = company;
        Salary = salary;
        Position = position;
        EmployeeId = employeeId;
        VacationDays = vacationDays;
        Department = department;
        YearsInCompany = yearsInCompany;
    }

    // Переопределение метода GetInfo
    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Департамент: {Department}, Должность: {Position}, Зарплата: {Salary:C}, ID: {EmployeeId}, Лет в компании: {YearsInCompany}, Отпускных дней: {VacationDays}";
    }

    // Переопределение метода SayHello
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте, меня зовут {Name}. Я работаю {Position} в {Company} уже {YearsInCompany} лет.");
    }

    // Метод для работы с другими сотрудниками
    public void WorkWith(Employee colleague)
    {
        Console.WriteLine($"{Name} работает вместе с {colleague.Name} над проектом в департаменте {Department}");
    }

    // Переопределяем метод взаимодействия для сотрудников
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Employee colleague && colleague.Company == Company)
        {
            Console.WriteLine($"{Name} взаимодействует с коллегой {colleague.Name} (ID: {colleague.EmployeeId}) из департамента {colleague.Department}");
            WorkWith(colleague);
        }
        else if (otherPerson is Student student)
        {
            Console.WriteLine($"{Name} проводит собеседование со студентом {student.Name} из {student.University}");
            Console.WriteLine($"{Name}: Расскажите о вашем опыте в {student.University}. Ваш средний балл {student.AverageGrade:F1} впечатляет!");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public void Work()
    {
        Console.WriteLine($"{Name} усердно работает на должности '{Position}'. Зарплата {Salary:C} не начисляется просто так!");
    }

    public void Work(int hours) // Перегрузка метода Work
    {
        Console.WriteLine($"{Name} усердно работает на должности '{Position}' {hours} часов в день.");
    }

    public void TakeVacation(int days)
    {
        if (days <= VacationDays)
        {
            VacationDays -= days;
            Console.WriteLine($"{Name} уходит в отпуск на {days} дней. Осталось отпускных дней: {VacationDays}.");
        }
        else
        {
            Console.WriteLine($"{Name}, у вас недостаточно отпускных дней. Запрошено: {days}, доступно: {VacationDays}.");
        }
    }

    // Новый метод для повышения зарплаты
    public virtual void RaiseSalary(decimal amount)
    {
        Salary += amount;
        Console.WriteLine($"{Name} получил(а) повышение зарплаты на {amount:C}. Новая зарплата: {Salary:C}");
    }

    // Перегрузка метода UpdateProfile
    public override void UpdateProfile(string email, string phone)
    {
        base.UpdateProfile(email, phone);
        Console.WriteLine($"(Сотрудник {Company})");
    }
}

public class Teacher : Person
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public List<Student> Students { get; set; }
    
    // --- ДОБАВЛЕННЫЕ АТРИБУТЫ ---
    public string TeacherId { get; set; }
    public string OfficeNumber { get; set; }
    public string AcademicDegree { get; set; } // Новый атрибут
    public int PublicationsCount { get; set; } // Новый атрибут

    // Конструктор с использованием свойств (ОБНОВЛЕН)
    public Teacher(string name, int age, string gender, string email, string phoneNumber, bool isMarried, string subject, int experience, string teacherId, string officeNumber, string academicDegree = "Кандидат наук", int publicationsCount = 0)
        : base(name, age, gender, email, phoneNumber, isMarried)
    {
        Subject = subject;
        Experience = experience;
        TeacherId = teacherId;
        OfficeNumber = officeNumber;
        AcademicDegree = academicDegree;
        PublicationsCount = publicationsCount;
        Students = new List<Student>();
    }

    // Переопределение метода SayHello
    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}. Я {AcademicDegree} и преподаю {Subject} уже {Experience} лет. Мой кабинет: {OfficeNumber}");
    }

    // Перегрузка метода SayHello
    public override void SayHello(string greetingType)
    {
        if (greetingType.ToLower() == "academic")
        {
            Console.WriteLine($"Добро пожаловать на лекцию! Я {AcademicDegree} {Name}, ваш преподаватель по {Subject}.");
        }
        else
        {
            base.SayHello(greetingType);
        }
    }

    // Метод для ответа на вопросы студентов
    public void AnswerQuestion(Student student, string question)
    {
        Console.WriteLine($"{Name} отвечает студенту {student.Name}: 'Хороший вопрос! Давайте разберем его на следующей лекции.'");
    }

    // Перегруженный метод AnswerQuestion
    public void AnswerQuestion(Student student, string question, bool isUrgent)
    {
        if (isUrgent)
        {
            Console.WriteLine($"{Name} немедленно отвечает студенту {student.Name}: 'Срочный вопрос! {question} - давайте обсудим это прямо сейчас в кабинете {OfficeNumber}.'");
        }
        else
        {
            AnswerQuestion(student, question);
        }
    }

    // Метод для добавления студента в список
    public void AddStudent(Student student)
    {
        Students.Add(student);
        Console.WriteLine($"{student.Name} добавлен(а) в список студентов преподавателя {Name}");
    }

    // Метод для проведения лекции
    public void ConductLecture()
    {
        Console.WriteLine($"\n{Name} проводит лекцию по {Subject} (ID преподавателя: {TeacherId})");
        foreach (var student in Students)
        {
            Console.WriteLine($"- {student.Name} (ID: {student.StudentId}) присутствует на лекции");
        }
    }

    // Перегруженный метод ConductLecture
    public void ConductLecture(string topic)
    {
        Console.WriteLine($"\n{Name} проводит лекцию по {Subject} на тему: '{topic}'");
        foreach (var student in Students)
        {
            Console.WriteLine($"- {student.Name} внимательно слушает про {topic}");
        }
    }

    // Переопределяем метод взаимодействия для преподавателей
    public override void InteractWith(Person otherPerson)
    {
        if (otherPerson is Student student)
        {
            Console.WriteLine($"{Name} взаимодействует со студентом {student.Name}");
            if (!Students.Contains(student))
            {
                AddStudent(student);
            }
            AnswerQuestion(student, "Общий вопрос по предмету");
        }
        else if (otherPerson is Teacher colleague)
        {
            Console.WriteLine($"{Name} обсуждает методики преподавания с {colleague.Name} (Кабинет: {colleague.OfficeNumber})");
            Console.WriteLine($"{Name}: Как вам новый учебный план по {Subject}?");
        }
        else
        {
            base.InteractWith(otherPerson);
        }
    }

    // --- ДОБАВЛЕННЫЕ МЕТОДЫ ---
    public void GradeAssignment(Student student, string course, double grade)
    {
        Console.WriteLine($"Преподаватель {Name} выставил(а) оценку {grade} студенту {student.Name} за курс '{course}'.");
        student.UpdateGrade(course, grade);
    }

    public void HoldConsultation()
    {
        Console.WriteLine($"Преподаватель {Name} проводит консультацию в кабинете {OfficeNumber}. Приглашаются все студенты курса {Subject}.");
    }

    // Перегруженный метод HoldConsultation
    public void HoldConsultation(string specialTopic)
    {
        Console.WriteLine($"Преподаватель {Name} проводит специальную консультацию по теме '{specialTopic}' в кабинете {OfficeNumber}.");
    }

    // Новый метод для публикации научной работы
    public void PublishResearch(string paperTitle)
    {
        PublicationsCount++;
        Console.WriteLine($"{Name} опубликовал(а) научную работу: '{paperTitle}'. Всего публикаций: {PublicationsCount}");
    }
}

// Основной код программы
Console.WriteLine("=== ДЕМОНСТРАЦИЯ РАБОТЫ РАСШИРЕННЫХ КЛАССОВ С ДЖЕНЕРИКАМИ ===\n");

// Создаем объекты с использованием ОБНОВЛЕННЫХ конструкторов
Student student1 = new Student("Артур", 19, "Мужской", "alshin06@gmail.com", "+79526759898", false, "ТИУ", 2, "S2024001", 4.2, "Информационные технологии", true);
Student student2 = new Student("Анна", 20, "Женский", "anna@university.ru", "+79098765432", false, "ТИУ", 3, "S2023005", 4.8, "Компьютерные науки");

Employee employee1 = new Employee("Иван", 35, "Мужской", "ivan@sber.ru", "+79501234567", true, "Сбербанк", 75000, "E1001", 21, "ИТ-отдел", 5, "Менеджер проектов");
Employee employee2 = new Employee("Елена", 28, "Женский", "elena@sber.ru", "+79507654321", false, "Сбербанк", 65000, "E1002", 18, "Аналитика", 2, "Старший аналитик");

Teacher teacher1 = new Teacher("Мария", 45, "Женский", "maria@university.ru", "+79215555555", true, "История", 13, "T050", "101A", "Доктор наук", 15);
Teacher teacher2 = new Teacher("Сергей", 50, "Мужской", "sergey@university.ru", "+79216666666", true, "Информатика", 20, "T025", "205B", "Профессор", 25);

// Демонстрация работы дженерик-класса
Console.WriteLine("=== РАБОТА С ДЖЕНЕРИК-РЕЕСТРОМ ===");
PersonRegistry<Person> registry = new PersonRegistry<Person>();

// Добавляем людей в реестр
registry.AddPerson(student1);
registry.AddPerson(student2);
registry.AddPerson(employee1);
registry.AddPerson(employee2);
registry.AddPerson(teacher1);
registry.AddPerson(teacher2);

// Поиск людей разными способами (демонстрация перегрузки)
Console.WriteLine("\n=== ПОИСК В РЕЕСТРЕ ===");
Person foundByName = registry.FindPerson("Артур");
Console.WriteLine($"Найден по имени: {foundByName?.GetInfo()}");

Person foundById = registry.FindPerson("S2023005", true);
Console.WriteLine($"Найден по ID: {foundById?.GetInfo()}");

// Статистика реестра
registry.PrintRegistryStats();

// Демонстрация перегрузки методов
Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ ===");
student1.SayHello();
student1.SayHello("student");
student1.SayHello("formal");

teacher1.SayHello();
teacher1.SayHello("academic");

employee1.Work();
employee1.Work(8);

teacher1.ConductLecture();
teacher1.ConductLecture("Архитектура древнего Рима");

teacher1.HoldConsultation();
teacher1.HoldConsultation("Подготовка к финальному экзамену");

// Демонстрация переопределения методов
Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПЕРЕОПРЕДЕЛЕНИЯ МЕТОДОВ ===");
Person[] people = new Person[] { student1, student2, employee1, employee2, teacher1, teacher2 };

foreach (var person in people)
{
    person.SayHello(); // Полиморфный вызов
    Console.WriteLine(person.GetInfo()); // Полиморфный вызов
    Console.WriteLine();
}

// Демонстрация новых методов
Console.WriteLine("=== ДЕМОНСТРАЦИЯ НОВЫХ МЕТОДОВ ===");
student1.ApplyForScholarship();
student2.ApplyForScholarship();

employee1.RaiseSalary(10000);
employee1.TakeVacation(5);

teacher1.PublishResearch("Новые подходы в преподавании истории");
teacher1.GradeAssignment(student1, "Программирование", 5.0);

// Демонстрация обновления профилей с перегрузкой
Console.WriteLine("\n=== ОБНОВЛЕНИЕ ПРОФИЛЕЙ ===");
student1.UpdateProfile("new_email@university.ru", "+79526750000");
employee1.UpdateProfile("ivan.new@sber.ru", "+79501230000", "Россия");

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ МЕЖДУ ОБЪЕКТАМИ ===");
student1.InteractWith(teacher1);
Console.WriteLine();

student1.InteractWith(student2);
Console.WriteLine();

employee1.InteractWith(employee2);
Console.WriteLine();

teacher1.InteractWith(student2);
Console.WriteLine();

employee1.InteractWith(student1);
Console.WriteLine();

teacher1.InteractWith(teacher2);
Console.WriteLine();

// Демонстрация перегруженного взаимодействия
Console.WriteLine("=== СПЕЦИАЛЬНЫЕ ВЗАИМОДЕЙСТВИЯ ===");
student1.InteractWith(teacher1, "учебный процесс");
teacher1.AnswerQuestion(student2, "Важный вопрос", true);
```
